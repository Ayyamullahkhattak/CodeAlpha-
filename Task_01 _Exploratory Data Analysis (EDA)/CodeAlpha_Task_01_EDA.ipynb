## **2. How many patients belong to the youth and adult groups?**

```python
# Create Age Groups covering the full range of ages (15 to 80)
bins = [14, 30, 80]
labels = ["Youth", "Adults"]

df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)

# Display the count of each group
print(df["Age_Group"].value_counts())
```

### **Insights**
The dataset contains 97,215 youth patients (24.30%) and 302,779 adult patients (75.70%). Therefore, adults represent the majority of patients in the dataset.

## **3. Which Cardiovascular risk factors are most common?**

```python
# Selecting Risk factor columns
risk_factors=["Hypertension", "Alcohol_Consumption", "Smoker","Blood_Sugar","Diabetes","Drug_Use"]
df["Blood_Sugar_Risk"] = np.where(
    df["Blood_Sugar"] >= 126,
    "High",
    "Normal"
)
```

```python
# Map the correct binary (Yes/No) risk factor columns in the dataset

risk_factors = ["Smoker", "Diabetes", "Hypertension", "Alcohol_Consumption", "Drug_Use"]

risk_counts = {}

for factor in risk_factors:
    # Using the loop variable 'factor' instead of the list 'risk_factors'
    risk_counts[factor] = (df[factor] == "Yes").sum()


    risk_counts["High_Blood_Sugar"] = (df["Blood_Sugar"] >= 126).sum()
# Convert to Series for prettier display
print(pd.Series(risk_counts).sort_values(ascending=False))
```

### **Insights**
High blood sugar is the most common cardiovascular risk factor in the dataset, affecting 228,460 patients (57.12%). This is followed by alcohol consumption (50.12%), smoking (50.09%),and diabetes (50.04%).


---

## **4. How do cardiovascular risk factors differ between youth and adults?**

```python
# Compare Blood Sugar risk (High vs Normal) between Youth and Adults
blood_sugar_comparison = pd.crosstab(df['Age_Group'], df['Blood_Sugar_Risk'], normalize='index') * 100
print("Blood Sugar Risk Factors in Youth and Adults:\n")
print(blood_sugar_comparison)
```

### **Insights**
Blood sugar risk was nearly identical between the two groups, with high blood sugar observed in 57.10% of youth and 57.13% of adults.


---

## **5. Is smoking more prevalent among youth or adults?**

```python
smoking_risk = pd.crosstab(
    df["Age_Group"],
    df["Smoker"],
    normalize="index"
) * 100

print(smoking_risk)
```

### **Insights**
Smoking is slightly more prevalent among youth (50.14%) than adults (50.0%).


---

## **6.How do blood pressure and cholesterol levels differ between the two age groups?**

```python
# Compare average blood pressure and cholesterol by age group

age_comparison = df.groupby("Age_Group")[["Blood_Pressure", "Cholesterol_Level"]].mean().round(2)

print(age_comparison)
```

### **Insights**
Youth have slightly higher average blood pressure (135.18) and cholesterol levels (200.19) compared with adults (134.89 and 200.10).


---

## **7. Does the type of chest pain differ between youth and adults?**

```python
chest_pain_comparison = pd.crosstab(
    df["Age_Group"],
    df["Chest_Pain"],
    normalize="index"
) * 100

print(chest_pain_comparison.round(2))
```

### **Insights**
Chest pain is slightly more common among youth (50.29%) than adults (49.92%).
